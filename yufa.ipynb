```python
def add_all(*numbers):
    return sum(numbers)

print(add_all(1, 2))
print(add_all(1, 2, 3, 4))
numbers = [1, 2, 3]

print(add_all(*numbers))
```

```python
words = ["apple", "hi", "banana", "cat"]

result = sorted(words ,key= lambda x : len(x) )

print(result)
```

```python
students = [
    {"name": "小明", "score": 85},
    {"name": "小红", "score": 92},
    {"name": "小刚", "score": 78}
]

result = sorted(
    students,
    key=lambda student: student["score"],
    reverse=True
)

for student in result:
    print(student["name"], student["score"])
```

```python
def calculate_average(*scores):
    return sum(scores)/len(scores)

print(calculate_average(80, 90, 100))
print(calculate_average(0))
```

```python
products = [
    {"name": "电脑", "price": 5000},
    {"name": "手机", "price": 3000},
    {"name": "耳机", "price": 500},
]

sorted(products, key=lambda product: product["price"])
```

```python
def log(func):
    def wrapper():
        print("函数开始")
        func()
        print("函数结束")

    return wrapper

@log
def say_hello():
    print("你好")

say_hello()
```
